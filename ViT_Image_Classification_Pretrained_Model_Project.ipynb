## 3. Introduction to Vision Transformer (ViT)

Vision Transformer is a Transformer-based architecture designed for computer vision tasks.

Unlike a traditional CNN that extracts features using convolutional filters, ViT divides an image into small patches. These patches are converted into embeddings and processed as a sequence by Transformer encoder layers.

### Basic Flow

```text
Input Image
     ↓
Resize / Normalize
     ↓
Image Patches
     ↓
Patch Embeddings
     ↓
Transformer Encoder
     ↓
Classification Head
     ↓
Class Scores
     ↓
Predicted Class
```

````python
# Import libraries
import os
import torch
import matplotlib.pyplot as plt
from PIL import Image
from transformers import ViTImageProcessor, ViTForImageClassification

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
````

## 8. Load the Pre-trained ViT Model

The following code downloads the model configuration, image processor, and pretrained model weights.

`model.eval()` places the model in evaluation mode because this project performs inference rather than training.

````python
# Load the pretrained Vision Transformer
model_name = "google/vit-base-patch16-224"

processor = ViTImageProcessor.from_pretrained(model_name)
model = ViTForImageClassification.from_pretrained(model_name)

model.eval()

print("Model loaded successfully.")
print("Model:", model_name)
print("Number of output classes:", model.config.num_labels)
````

## 9. Understanding the Model

The model contains:
- A patch-based image representation
- Transformer encoder layers
- A classification head
- Learned pretrained weights

The pretrained weights are the numerical parameters learned during the model's original training. During inference, these weights are used to transform an input image into classification scores.

````python
# Display basic model information
print("Architecture:", model.config.architectures)
print("Hidden size:", model.config.hidden_size)
print("Number of attention heads:", model.config.num_attention_heads)
print("Number of transformer layers:", model.config.num_hidden_layers)
print("Image size:", model.config.image_size)
print("Patch size:", model.config.patch_size)
````

## 10. Input Preparation

Upload or place a test image in the same folder as this notebook.

Recommended examples:
- dog.jpg
- cat.jpg
- car.jpg
- bird.jpg
- banana.jpg

The image should contain a clear object that belongs to one of the model's supported classes.

````python
# Change this path to your image file
image_path = "dog.jpg"

if not os.path.exists(image_path):
    raise FileNotFoundError(
        f"Image not found: {image_path}. "
        "Place the image in the notebook folder or change image_path."
    )

image = Image.open(image_path).convert("RGB")

plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.title("Input Image")
plt.axis("off")
plt.show()
````

## 11. Pre-processing

The ViT image processor performs the preprocessing required by the model, including resizing, normalization, and conversion into tensors.

The processed image is then passed to the model.

````python
# Preprocess the image
inputs = processor(images=image, return_tensors="pt")

print("Input tensor shape:", inputs["pixel_values"].shape)
````

## 12. Model Inference

`torch.no_grad()` is used because the model is not being trained. It prevents unnecessary gradient calculations and reduces memory usage during inference.

````python
# Perform inference
with torch.no_grad():
    outputs = model(**inputs)

logits = outputs.logits

print("Logits shape:", logits.shape)
````

## 13. Convert Logits to Probabilities

The model produces logits. Softmax converts these scores into probabilities.

The class with the highest probability is selected as the primary prediction.

````python
# Convert logits to probabilities
probabilities = torch.softmax(logits, dim=-1)

predicted_index = probabilities.argmax(dim=-1).item()
predicted_label = model.config.id2label[predicted_index]
confidence = probabilities[0, predicted_index].item()

print("Predicted class:", predicted_label)
print(f"Confidence: {confidence * 100:.2f}%")
````

## 14. Top-5 Predictions

Displaying the top-5 predictions provides more information than showing only one class.

````python
# Get top-5 predictions
top_k = 5
top_values, top_indices = torch.topk(probabilities, top_k)

print("Top-5 Predictions")
print("-" * 40)

for rank, (score, index) in enumerate(zip(top_values[0], top_indices[0]), start=1):
    label = model.config.id2label[index.item()]
    print(f"{rank}. {label}: {score.item() * 100:.2f}%")
````

## 15. Result Visualization

The following cell displays the input image together with the predicted class and confidence score.

````python
# Display final classification result
plt.figure(figsize=(7, 7))
plt.imshow(image)
plt.title(f"Prediction: {predicted_label}\nConfidence: {confidence * 100:.2f}%")
plt.axis("off")
plt.show()
````

## 16. Reusable Image Classification Function

The following function allows multiple images to be tested without repeating the complete inference code.

````python
def classify_image(image_path, show_top5=True):
    """Classify one image using the pretrained ViT model."""

    if not os.path.exists(image_path):
        print(f"File not found: {image_path}")
        return None

    image = Image.open(image_path).convert("RGB")
    inputs = processor(images=image, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=-1)
    top_values, top_indices = torch.topk(probabilities, 5)

    predicted_index = top_indices[0, 0].item()
    predicted_label = model.config.id2label[predicted_index]
    confidence = top_values[0, 0].item()

    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.title(f"{predicted_label} - {confidence * 100:.2f}%")
    plt.axis("off")
    plt.show()

    if show_top5:
        print("Top-5 predictions:")
        for rank, (score, index) in enumerate(
            zip(top_values[0], top_indices[0]), start=1
        ):
            label = model.config.id2label[index.item()]
            print(f"{rank}. {label}: {score.item() * 100:.2f}%")

    return predicted_label, confidence
````

Predicted Class
